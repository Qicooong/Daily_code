```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time
```

```python
def fetch_etf_data():

    # 初始化 WebDriver
    service = webdriver.edge.service.Service("C:/msedgedriver.exe")  # 替换为 EdgeDriver 路径
    driver = webdriver.Edge(service=service)
    url = "https://www.sse.com.cn/disclosure/diclosure/block/deal/dzjyxx/"
    driver.get(url)
    driver.maximize_window()

    try:
        WebDriverWait(driver, 20).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, ".sse_outerItem.js_dateRange"))
        )
        print("页面加载完成")

 
        print("设置时间范围...")
        date_range_button = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.CSS_SELECTOR, ".sse_outerItem.js_dateRange input.form-control"))
        )
        date_range_button.click()
        time.sleep(1)
        driver.execute_script("""
            document.querySelectorAll('.sse_outerItem.js_dateRange input.form-control')[0].value = '2024-10-13 - 2025-01-13';    
        """)      # 设置时间
        time.sleep(1)
        confirm_button = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.CSS_SELECTOR, ".laydate-footer-btns .laydate-btns-confirm"))
        )
        confirm_button.click()
        time.sleep(5)  # 等待页面刷新
        print("时间范围设置成功")


        print("开始抓取数据...")
        all_data = []
        etf_data = []
        page = 1

        while True:
            print(f"正在抓取第 {page} 页的数据...")

            time.sleep(3)
            rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")

            if not rows:
                print("当前页面无数据，结束抓取")
                break

            for row in rows:
                cols = row.find_elements(By.TAG_NAME, "td")
                if len(cols) == 9:
                    row_data = [col.text.strip() for col in cols]
                    all_data.append(row_data)
                    if "ETF" in row_data[1]:  # 判断是否是 ETF 数据
                        etf_data.append(row_data)
                else:
                    print(f"跳过列数不匹配的行：{[col.text.strip() for col in cols]}")

            # 下一页逻辑
            try:
                next_button = driver.find_element(By.CSS_SELECTOR, ".pagination-box .next a")
                if "disabled" in next_button.get_attribute("class"):
                    print("已到最后一页")
                    break
                next_button.click()
                page += 1
                time.sleep(3)
            except Exception as e:
                print("无法找到下一页按钮或翻页失败，结束抓取")
                break

        driver.quit()

        # 保存数据
        print("正在保存数据...")
        if all_data:
            columns = ["交易日期", "证券简称", "证券代码", "成交价格(元)", "成交金额(万元)", "成交量(手)", "买入营业部", "卖出营业部", "是否专场"]
            df_all = pd.DataFrame(all_data, columns=columns)
            df_all.to_csv("all_data.csv", index=False, encoding="utf-8")
            print("所有数据已保存到 all_data.csv 文件中")

            if etf_data:
                df_etf = pd.DataFrame(etf_data, columns=columns)
                df_etf.to_csv("etf_data.csv", index=False, encoding="utf-8")
                print("ETF 数据已保存到 etf_data.csv 文件中")
            else:
                print("未找到任何 ETF 数据")
        else:
            print("未找到任何数据")

    except Exception as e:
        print(f"页面加载或操作过程中出错: {e}")
        driver.quit()


fetch_etf_data()
```
